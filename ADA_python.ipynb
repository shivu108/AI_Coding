```python
import pandas as pd

operators = ["==", "!=", ">", "<", ">=", "<="]

meaning = [
    "Equal To",
    "Not Equal To",
    "Greater Than",
    "Less Than",
    "Greater Than or Equal To",
    "Less Than or Equal To"
]

df = pd.DataFrame({
    "Meaning": meaning,
    "Operator": operators
})

df.style.set_properties(
    subset=["Meaning"],
    **{"text-align": "left"}
).set_table_styles([
    {
        "selector": "th.col_heading",
        "props": [("text-align", "left")]
    }
])
```

```python
# This will make:

# Meaning heading → left aligned
# Meaning values → left aligned
# Operator → unchanged
```

```python
# 👍 If you are using a normal .py file in VS Code, you don't need .style or jinja2.

# Use Python's ljust() to control the alignment:
```

```python
import pandas as pd

operators = ["==", "!=", ">", "<", ">=", "<="]

meaning = [
    "Equal To",
    "Not Equal To",
    "Greater Than",
    "Less Than",
    "Greater Than or Equal To",
    "Less Than or Equal To"
]

df = pd.DataFrame({
    "Meaning": meaning,
    "Operator": operators
})

print("Meaning".ljust(30), "Operator")

for i in range(len(df)):
    print(df["Meaning"][i].ljust(30), df["Operator"][i])
```

```python
# Output:

# Meaning                        Operator
# Equal To                       ==
# Not Equal To                   !=
# Greater Than                   >
# Less Than                      <
# Greater Than or Equal To       >=
# Less Than or Equal To          <=
```

```python
# What does ljust(30) mean?

# "Meaning".ljust(30)

# Keep the text left-aligned and give it a width of 30 spaces.

print("Hello".ljust(20), "World")
print("Hello".ljust(30), "World")
```
